```python
journey_data = journey_data.sort_values(["user_id", "timestamp"])
```

```python
journey_data["touchpoint_order"] = journey_data.groupby("user_id").cumcount() + 1
```

```python
journey_data.head()
```

```python
journey_data["total_touchpoints"] = journey_data.groupby("user_id")["user_id"].transform("count")
```

```python
journey_data.head()
```

```python
journey_data["decay_weight"] = (
    journey_data["touchpoint_order"] /
    journey_data["total_touchpoints"]
)
```

```python
journey_data.head()
```

```python
journey_data["weight_sum"] = journey_data.groupby("user_id")["decay_weight"].transform("sum")
```

```python
journey_data.head()
```

```python
journey_data["weight_sum"] = journey_data.groupby("user_id")["decay_weight"].transform("sum")
```

```python
journey_data.head()
```

```python
journey_data["time_decay_credit"] = (
    journey_data["revenue"] *
    (journey_data["decay_weight"] / journey_data["weight_sum"])
)
```

```python
journey_data.head()
```

```python
journey_data.groupby("channel")["time_decay_credit"].sum()
```

```python
time_decay_attribution = journey_data.groupby("channel")["time_decay_credit"].sum()
```

```python
comparison = pd.DataFrame({
    "First Touch": first_touch.groupby("channel")["revenue"].sum(),
    "Last Touch": last_touch.groupby("channel")["revenue"].sum(),
    "Linear Attribution": linear_attribution,
    "Time Decay Attribution": time_decay_attribution
})
```

Project Conclusion

This project analyzed customer journeys using multiple attribution models.

First Touch, Last Touch, Linear Attribution, and Time Decay Attribution were compared to understand channel influence across the customer journey.

Results demonstrated that channel performance varies depending on the attribution model selected.

Time Decay Attribution was identified as the most practical model for decision-making because it reflects the influence of touchpoints closer to conversion while still considering the full customer journey.
