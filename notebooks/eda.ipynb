```python
journey_data = web_analytics.merge(
    crm_conversions,
    on="user_id",
    how="inner"
)
```

```python
journey_data.head()
```

```python
journey_data.shape
```

```python
journey_data["channel"].value_counts()
```

```python
journey_data["channel"].value_counts().plot(kind="bar")
plt.title("Conversions by Marketing Channel")
plt.show()
```

```python
journey_data.groupby("channel")["revenue"].sum().sort_values(ascending=False)
```

```python
journey_data.groupby("channel")["revenue"].mean().sort_values(ascending=False)
```

```python
first_touch = journey_data.sort_values("timestamp").groupby("user_id").first()
```

```python
first_touch.head()
```

```python
first_touch["channel"].value_counts()
```

```python
first_touch.groupby("channel")["revenue"].sum().sort_values(ascending=False)
```

```python
last_touch = journey_data.sort_values("timestamp").groupby("user_id").last()
```

```python
last_touch["channel"].value_counts()
```
